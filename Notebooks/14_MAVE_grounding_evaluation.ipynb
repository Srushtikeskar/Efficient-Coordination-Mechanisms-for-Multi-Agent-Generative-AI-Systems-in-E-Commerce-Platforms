### MAVE 

```python
# =====================================================
# 01. IMPORTS AND PROJECT PATHS
# =====================================================

from pathlib import Path
import sys
import re
import json
import ast

import numpy as np
import pandas as pd


# =====================================================
# PROJECT ROOT
# =====================================================

PROJECT_ROOT = Path.cwd()

if not (PROJECT_ROOT / "Agents").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent


if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))


MAVE_FOLDER = (
    PROJECT_ROOT
    / "Data"
    / "Raw"
    / "mave"
)

RESULTS_FOLDER = (
    PROJECT_ROOT
    / "Results"
)

RESULTS_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)


print("PROJECT_ROOT:", PROJECT_ROOT)
print("MAVE_FOLDER:", MAVE_FOLDER)
print("RESULTS_FOLDER:", RESULTS_FOLDER)
```

Locate MAVE files

```python
# =====================================================
# 02. LOCATE MAVE FILES
# =====================================================

positive_path = (
    MAVE_FOLDER
    / "mave_positives_labels_sample.jsonl"
)

negative_path = (
    MAVE_FOLDER
    / "mave_negatives_labels_sample.jsonl"
)


print(
    "Positive exists:",
    positive_path.exists()
)

print(
    "Negative exists:",
    negative_path.exists()
)


if not positive_path.exists():
    raise FileNotFoundError(
        positive_path
    )

if not negative_path.exists():
    raise FileNotFoundError(
        negative_path
    )
```

```python
# =====================================================
# 03. LOAD MAVE
# =====================================================

mave_positive = pd.read_json(
    positive_path,
    lines=True
)

mave_negative = pd.read_json(
    negative_path,
    lines=True
)


print(
    "Positive shape:",
    mave_positive.shape
)

print(
    "Negative shape:",
    mave_negative.shape
)


print("\nPositive columns:")
print(
    mave_positive.columns.tolist()
)

print("\nNegative columns:")
print(
    mave_negative.columns.tolist()
)


print("\nPOSITIVE EXAMPLES")
display(
    mave_positive.head(3)
)


print("\nNEGATIVE EXAMPLES")
display(
    mave_negative.head(3)
)
```
