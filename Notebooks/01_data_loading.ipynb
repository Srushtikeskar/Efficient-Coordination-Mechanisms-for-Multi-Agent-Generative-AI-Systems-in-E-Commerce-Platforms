## Loading datasets


```python
#importing necessary libraries

import pandas as pd
from pathlib import Path
import numpy as np
```

```python
#displaying all columns

pd.set_option("display.max_columns", None)

pd.set_option("display.max_colwidth", None)
```

Loading Metadata Datset

```python
project_path = Path.cwd().parent
raw_data_path = project_path/"Data"/"Raw"

print(raw_data_path)
```

```python
metadata = pd.read_parquet(
    raw_data_path/"Raw_meta_Amazon_Cellphones.parquet"
    
)

print("Metadata Loaded Successfully!")
```

Loading Reviews Data

```python
reviews = pd.read_parquet(
    raw_data_path/"Raw_review_Amazon_Cellphones.parquet"
    
)

print("Reviews Loaded Successfully!")
```

```python
##Checking duplicates
metadata_duplicates = metadata["parent_asin"].duplicated().sum()
print(f"Duplicate parent_asin in Metdata: {metadata_duplicates}")
```

```python
review_duplicates = reviews.duplicated(
    subset=["parent_asin", "user_id","timestamp"]).sum()

print(f"Duplicate Reviews: {review_duplicates}")
```

```python
#verifying the common key

print("Metadata parent_asin Exists:", "parent_asin" in metadata.columns)
print("Reviews parent_asin Exists:", "parent_asin" in reviews.columns)
```

```python
# =====================================================
# SAVE PARQUET DATASETS AS CSV FILES
# =====================================================

# Folder where CSV files will be saved
csv_data_path = project_path / "Data" / "CSV"

# Create folder if it does not already exist
csv_data_path.mkdir(parents=True, exist_ok=True)


# -----------------------------------------------------
# Save Metadata
# -----------------------------------------------------

metadata.to_csv(
    csv_data_path / "Amazon_Cellphones_Metadata.csv",
    index=False
)

print("Metadata CSV saved successfully!")


# -----------------------------------------------------
# Save Reviews
# -----------------------------------------------------

reviews.to_csv(
    csv_data_path / "Amazon_Cellphones_Reviews.csv",
    index=False
)

print("Reviews CSV saved successfully!")


# -----------------------------------------------------
# Display saved locations
# -----------------------------------------------------

print("\nFiles saved at:")
print(csv_data_path / "Amazon_Cellphones_Metadata.csv")
print(csv_data_path / "Amazon_Cellphones_Reviews.csv")
```
