## Cleaning and processing datasets

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
project_path = Path.cwd().parent
raw_data_path = project_path/"Data"/"Raw"

# print(raw_data_path)

metadata = pd.read_parquet(
    raw_data_path/"Raw_meta_Amazon_Cellphones.parquet"
    
)

print("Metadata Loaded Successfully!")

reviews = pd.read_parquet(
    raw_data_path/"Raw_review_Amazon_Cellphones.parquet"
    
)

print("Reviews Loaded Successfully!")

print(metadata.shape)
print(reviews.shape)
```

```python
metadata_clean = metadata.copy()
reviews_clean = reviews.copy()
```

-- Metadata has bought_together, author, subtitle, store and main_category nulls value columns.
-- Ratings has no null value column

```python
# Droping unneeded columns 

metadata_clean = metadata_clean.drop (
    columns = ["bought_together",
        "author",
        "subtitle",
        ]
)
```

```python
#filling missing values

metadata_clean["store"]=metadata_clean["store"].fillna("Unknown")
metadata_clean["main_category"] = metadata_clean["main_category"].fillna("Cell Phones & Accessories")

```

Removing duplicates

```python
##Checking duplicates
metadata_duplicates = metadata_clean["parent_asin"].duplicated().sum()
print(f"Duplicate parent_asin in Metdata: {metadata_duplicates}")
```

```python
review_duplicates = reviews_clean.duplicated(
    subset=[
        "parent_asin",
        "user_id",
        "timestamp",
        "text"
    ]
).sum()

print(f"Duplicate Review Records Before Removal: {review_duplicates}")
```

```python
reviews_clean = reviews_clean.drop_duplicates(
    subset=[
        "parent_asin",
        "user_id",
        "timestamp",
        "text"
    ]
).reset_index(drop=True)
```

```python
review_duplicates_after = reviews_clean.duplicated(
    subset=[
        "parent_asin",
        "user_id",
        "timestamp",
        "text"
    ]
).sum()

print(f"Duplicate Review Records After Removal: {review_duplicates_after}")
```

```python
#converting metadata price to float

metadata_clean["price"] = (
    metadata_clean["price"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

metadata_clean["price"] = pd.to_numeric(
    metadata_clean["price"],
    errors="coerce"
)
```

```python
#converting reviews timestamp to datetime


reviews_clean["review_date"] = pd.to_datetime(
    reviews_clean["timestamp"],
    unit="ms"
)


```

```python

# print(reviews_clean["images"].iloc[0])
# print(type(reviews_clean["images"].iloc[0]))

#converting ireview images to python list


reviews_clean["images"] = reviews_clean["images"].apply(
    lambda x: x.tolist() if isinstance(x, np.ndarray) else x
)
```

```python
#converting dictionaries to python lists for better processing

def convert_nested(obj):
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    elif isinstance(obj, dict):
        return {k: convert_nested(v) for k, v in obj.items()}
    elif isinstance(obj, list):
        return [convert_nested(v) for v in obj]
    else:
        return obj

metadata_clean["images"] = metadata_clean["images"].apply(convert_nested)
metadata_clean["videos"] = metadata_clean["videos"].apply(convert_nested)
```

```python
# for null pricing : 

title_price_counts = (
    metadata_clean.groupby("title")["price"]
    .nunique(dropna=True)
)

consistent_titles = title_price_counts[title_price_counts == 1].index
```

```python
price_lookup = (
    metadata_clean[metadata_clean["title"].isin(consistent_titles)]
    .groupby("title")["price"]
    .first()
)

mask = metadata_clean["price"].isna()

metadata_clean.loc[mask, "price"] = (
    metadata_clean.loc[mask, "title"]
    .map(price_lookup)
)
```

```python
text_cols = [
    "title",
    "description",
    "features",
    "categories",
    "details"
]

for col in text_cols:
    metadata_clean[col] = metadata_clean[col].str.strip()
```

```python
text_cols = ["title", "text"]

for col in text_cols:
    reviews_clean[col] = reviews_clean[col].str.strip()
```

creating a product_text column in metadata which the Retrieval Agent will embedd during FAISS

```python
metadata_clean["product_text"] = (
    "Title: " + metadata_clean["title"].fillna("") +
    "\n\nDescription: " + metadata_clean["description"].fillna("") +
    "\n\nFeatures: " + metadata_clean["features"].fillna("") +
    "\n\nCategory: " + metadata_clean["categories"].fillna("") +
    "\n\nDetails: " + metadata_clean["details"].fillna("")
)
```

Saving the cleaned datasets


```python
from pathlib import Path

# Create Processed folder if it doesn't exist
processed_path = Path("../Data/Cleaned")
processed_path.mkdir(parents=True, exist_ok=True)

# Save datasets
metadata_clean.to_parquet(
    processed_path / "metadata_clean.parquet",
    index=False
)

reviews_clean.to_parquet(
    processed_path / "reviews_clean.parquet",
    index=False
)

print("Datasets saved successfully!")
```
