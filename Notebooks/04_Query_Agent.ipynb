### QUERY AGENT

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from pydantic import BaseModel,Field



load_dotenv()
```

```python



class ProductQuery(BaseModel):

    product_type: str = Field(
        description="The type of product requested by the user"
    )

    brand: str | None = Field(
        default=None,
        description="The requested brand, or null if not mentioned"
    )

    budget: float | None = Field(
        default=None,
        description=(
            "Maximum budget as a numeric value only. "
            "Return null if no budget is mentioned."
        )
    )

    features: list[str] = Field(
        default_factory=list,
        description="Important features explicitly requested by the user"
    )
```

```python
class QueryAgent:

    def __init__(self):

        self.llm = ChatOpenAI(
            model="gpt-4o-mini",
            temperature=0
        )

        self.structured_llm = self.llm.with_structured_output(
            ProductQuery
        )

    def parse(self, user_query):

        if not user_query.strip():
            raise ValueError(
                "User query cannot be empty."
            )

        return self.structured_llm.invoke(user_query)
```

```python
query_agent = QueryAgent()
```

```python
query = """
I need an apple phone lowest price
with good camera and long battery life.
"""

response = query_agent.parse(query)

print("Product Type :", response.product_type)
print("Brand        :", response.brand)
print("Budget       :", response.budget)
print("Features     :", response.features)
```

```python
queries = [

    "Samsung phone under 30000",

    "USB C charger for Samsung",

    "Wireless earbuds with noise cancellation",

    "Gaming phone with Snapdragon processor",

    "Phone with AMOLED display and 5000mAh battery",

    "iPhone with good camera",

    "Fast charging power bank",

    "Bluetooth speaker under 5000"

]

for query in queries:

    print("=" * 80)

    print("USER QUERY:")
    print(query)

    result = query_agent.parse(query)

    print()

    print(result)
```
