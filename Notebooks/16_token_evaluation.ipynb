### TOKEN EVALUATION

```python


from pathlib import Path
import pandas as pd


# -----------------------------------------------------
# PROJECT PATH
# -----------------------------------------------------

PROJECT_ROOT = Path("..").resolve()

RESULTS_FOLDER = (
    PROJECT_ROOT
    / "Results"
)


# -----------------------------------------------------
# LOAD TOKEN RESULTS
# -----------------------------------------------------

single_tokens = pd.read_csv(
    RESULTS_FOLDER
    / "single_agent_token_usage.csv"
)

no_coord_tokens = pd.read_csv(
    RESULTS_FOLDER
    / "no_coord_token_usage.csv"
)

coordinated_tokens = pd.read_csv(
    RESULTS_FOLDER
    / "coordinated_token_usage.csv"
)


print(
    "Single Agent rows:",
    len(single_tokens)
)

print(
    "Without Coordination rows:",
    len(no_coord_tokens)
)

print(
    "Coordinated rows:",
    len(coordinated_tokens)
)


# -----------------------------------------------------
# COMBINE
# -----------------------------------------------------

all_token_results = pd.concat(
    [
        single_tokens,
        no_coord_tokens,
        coordinated_tokens
    ],
    ignore_index=True
)


print(
    "\nTotal runs:",
    len(all_token_results)
)


# -----------------------------------------------------
# CHECK FOR MISSING VALUES
# -----------------------------------------------------

print(
    "\nMissing token values:"
)

print(
    all_token_results[
        [
            "prompt_tokens",
            "completion_tokens",
            "total_tokens"
        ]
    ].isna().sum()
)


# -----------------------------------------------------
# TOKEN EFFICIENCY SUMMARY
# -----------------------------------------------------

token_summary = (
    all_token_results
    .groupby(
        "system"
    )
    .agg(

        queries=(
            "test_id",
            "count"
        ),

        avg_prompt_tokens=(
            "prompt_tokens",
            "mean"
        ),

        avg_completion_tokens=(
            "completion_tokens",
            "mean"
        ),

        avg_total_tokens=(
            "total_tokens",
            "mean"
        ),

        median_total_tokens=(
            "total_tokens",
            "median"
        ),

        min_total_tokens=(
            "total_tokens",
            "min"
        ),

        max_total_tokens=(
            "total_tokens",
            "max"
        ),

        total_tokens=(
            "total_tokens",
            "sum"
        ),

        avg_cost_usd=(
            "estimated_cost_usd",
            "mean"
        ),

        total_cost_usd=(
            "estimated_cost_usd",
            "sum"
        )

    )
    .reset_index()
)


# -----------------------------------------------------
# ROUND RESULTS
# -----------------------------------------------------

token_summary[
    [
        "avg_prompt_tokens",
        "avg_completion_tokens",
        "avg_total_tokens",
        "median_total_tokens",
        "avg_cost_usd",
        "total_cost_usd"
    ]
] = token_summary[
    [
        "avg_prompt_tokens",
        "avg_completion_tokens",
        "avg_total_tokens",
        "median_total_tokens",
        "avg_cost_usd",
        "total_cost_usd"
    ]
].round(6)


display(
    token_summary
)
```
